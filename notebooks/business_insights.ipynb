Objective: Convert SQL analysis and Power BI findings into business insights for insurace management, focusing on premium collection, claims exposure, customer risk, and product portfolio performance.

This is a small illustrative dataset  (10 customers, 15 policies, 15 payments and 15 claims). Findings are descriptive and should not be treated as statistically generalizable or actuarial conclusions.

### 1. Executive Summary

The portfolio has written premiums of ₹193,500 and collected ₹166,500, resulting in a premium collection rate of 86.05%. Outstanding premiums total ₹1,245,000 with ₹942,000 approved. The rule-based customer risk segmentation identifies 4 High Risk, 2 Medium Risk, and 4 Low Risk customers.

Management priorities suggested by these results:

1) Improve follow-up on pending and failed premium payments.
2) Review high claim exposure in Life and Health products.
3) Investigate claim rejection and pending patterns, particularly in Motor and customers with repeated claims.
4) Treat customer risk scores as screening indicators, not validated underwriting or fraud scores.

Insight: Life has the largest approved claim amount(₹530,000), followed by Health(₹305,000). Health and Life also show the highest indicative approved-claims-to-collected-premium rates.

Business implication: Conduct a deeper profitability review of Life and Health portfolios using exposure-adjusted, earned-premium, and incurred-claims data before making pricing or portfolio decisions.

### 6. Key Business Findings.

1) Collection performance: Premium collection rate is 86.05%, with ₹27,000 outstanding.
2) Collection timing: March leads monthly collections at ₹47,000. May's low value needs a completeness check before trend conclusions.
3) Claims exposure: ₹9.42 lakh in claims are approved; claims approval rate is 73.33%.
4) Product concentration: Life and Health contribute the largest approved claim amounts and warrant deeper profitability investigation.
5) Customer concentration: Four customers are High Risk under the project's rule-based score; five have multiple claims.
6) Actionable opportunity: Improve payment follow-up and investigate claim status / rejection patterns, while validating the risk logic with more data.

### Conclusion

The analysis highlights opportunities to improve premium collections, strengthen claims monitoring, and focus management attention on higher-exposure product categories and customers. The dashboard is a decision-support tool: further data and validation are required before treating the risk scores or product ratios as predictive, actuarial or causal measures.



```python
# Optional : Recreate key portfolio charts
import pandas as pd
import matplotlib.pyplot as plt 

monthly_collections = pd.DataFrame({
    'Month': ['January', 'February', 'March', 'April', 'May'],
    'Collected Premium': [38000, 40000, 47000, 36000, 5500]
})

monthly_collections.plot(x='Month', y='Collected Premium', kind='bar', legend=False)
plt.title('Monthly Premium Collections')
plt.ylabel('Collected Premiun (₹)')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Risk segment distribution
risk_distribution = pd.DataFrame({
    'Risk Category': ['High Risk', 'Medium Risk', 'Low Risk'],
    'Customers': [4, 2, 4]
})
risk_distribution.plot(x='Risk Category', y='Customers', kind='bar', legend=False)
plt.title('Customer Risk Segment Distribution')
plt.ylabel('Number of Customers')
plt.tight_layout()
plt.show()
```

```python
# Approved claims by product category
product_claims = pd.DataFrame({
    'Insurance Type': ['Health', 'Life', 'Motor', 'Travel'],
    'Approved Claims': [305000, 530000, 75000, 32000]
})
product_claims.plot(x='Insurance Type', y='Approved Claims', kind='bar', legend=False)
plt.title('Approved Claim Amount by Insurance Type')
plt.ylabel('Approved Claim Amount (₹)')
plt.tight_layout()
plt.show()
```
