#### 7. Limitations & Interpretation Cautions

- This is a small sample dataset (10 customers and 15 policies) and results should not be generalized to be a real insurer's full portfolio.
- Gender and signup date are missing in the source data.
- The dataset does not include earned premium, reserves, full policy exposure periods, claim settlement dates, or actual payout records.
- Approved claim amounts are not confirmed actual cash payouts.
- Claims-to-collected-premium ratios are indicative only and must not be presented as formal loss ratios.
- Customer risk categories are rule-based and exploratory; they have not been statistically validated.They should support review, not automated adverse decisions.
